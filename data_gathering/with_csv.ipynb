## 14. Converters

```python
def converter(name):
    if name=='First Owner':
        return 1
    elif name=='Second Owner':
        return 2
    elif name=='Third Owner	':
        return 3
    else:
        return 4
```

```python
pd.read_csv('/Users/kishan/code/genai/ml/ml_learning/datasets/cardekho.csv',converters={'owner':converter})
```
