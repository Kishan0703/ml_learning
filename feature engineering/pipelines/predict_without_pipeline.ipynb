```python
import pickle
import numpy as np
```

```python
ohe_sex= pickle.load(open('models/one_sex.pkl', 'rb'))
ohe_embarked= pickle.load(open('models/one_embarked.pkl', 'rb'))
clf= pickle.load(open('models/clf.pkl', 'rb'))
```

```python
test_input= np.array([[3, 1, 0, 0, 7.25, 27.0, 'S']], dtype=object)  
```

```python
test_input_sex= ohe_sex.transform(test_input[:,1].reshape(1, 1))
test_input_embarked= ohe_embarked.transform(test_input[:, -1].reshape(1, 1))
```

```python
test_input_sex
```

```python
test_input_age= test_input[:, 5].astype(float).reshape(1, 1)
```

```python
test_input_transformed= np.hstack((
    test_input[:, [0,2,3,4]].astype(float),
    test_input_age,
    test_input_sex,
    test_input_embarked
))
```
